# Stellar Diagrams Tutorial
This tutorial demonstrates how to fetch data from Gaia DR3 and the TESS Input Catalog (TIC) to plot Color-Magnitude Diagrams (CMD) and Hertzsprung-Russell (HR) diagrams.

```python
from stellar_diagrams import StellarDiagramMaker

# Disable warnings for a cleaner notebook output
import warnings
warnings.filterwarnings("ignore")
```

## Example 1: Gaia DR3 Color-Magnitude Diagram (Hexbin Plot)
Querying 25,000 stars up to an apparent magnitude of 11.

```python
# Initialize for Gaia
gaia_maker = StellarDiagramMaker(catalog='gaia', max_stars=1000, mag_limit=10)

# Fetch and clean the data
gaia_data = gaia_maker.fetch_data()

# Plot a CMD using a hexbin plot (great for dense datasets)
_ = gaia_maker.plot(diagram='cmd', style='hex')
```

## Example 2: Gaia HR Diagram (Kernel Density Estimation)
Visualizing the same Gaia data, but mapped as Temperature vs Luminosity.

```python
# Use the previously fetched data to plot an HR diagram with KDE styling
_ = gaia_maker.plot(diagram="hr", style="scatter")
```

## Example 3: TIC HR Diagram (Scatter Plot)
Querying the TESS Input Catalog for a smaller sample of stars.

```python
# Initialize for TIC (Note: MAST queries can take longer than Gaia ADQL)
tic_maker = StellarDiagramMaker(catalog='tic', max_stars=5000, mag_limit=10)

# Fetch the data
tic_data = tic_maker.fetch_data()

# Plot an HR diagram using a standard scatter plot
tic_maker.plot(diagram='hr', style='scatter')
```
